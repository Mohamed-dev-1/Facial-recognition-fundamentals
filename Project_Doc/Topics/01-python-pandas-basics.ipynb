<h4> <b> 1. Reading data :</b></h4>

* **pd.read_csv("file.csv")** For csv files
* **pd.read_excel("file.xlsx")** For Excel files
* **pd.read_json("file.json")** For json files


All methods loads the file into a Dataframe and a Dataframe is simply a table-like structure of rows and named columns .
